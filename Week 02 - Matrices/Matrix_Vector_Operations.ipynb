```python
X = df[numeric_columns].to_numpy()
print(X.shape)
X[:5]
```

## Part 1: Basic Matrix and Vector Operations

### 1. Matrix Addition

We'll take two small batches of rows from `X` and add them together. Adding two batches is like combining two groups of people into one running total.

```python
batch_a = X[:5]
batch_b = X[5:10]

batch_sum = batch_a + batch_b
print(batch_sum)
```

### 2. Matrix Subtraction

Subtracting shows the difference between two batches — useful for comparing groups (e.g., "how does this batch differ from that one?").

```python
batch_diff = batch_a - batch_b
print(batch_diff)
```

### 3. Matrix Multiplication

- **Multiplying by an identity matrix** — a special matrix that changes nothing — proves multiplication works as expected.
- **Multiplying by a scaling matrix** shrinks or grows each column by a chosen amount. This is exactly what "feature scaling" means in AI: making columns comparable so no single one dominates a model.

```python
identity = np.eye(6)
batch_a_identity = batch_a @ identity
print(batch_a_identity)
print('\n')

scaling = np.diag([2, 1, 1, 1, 1,  1])
batch_a_scale = batch_a_identity @ scaling
print(batch_a_scale)
```

### 4. Dot Product

The dot product combines two lists of numbers (vectors) into a single number. It's how AI models measure how "similar" or "aligned" two things are.

```python
batch_dot = np.dot(X[:, 0], X[:, 5])
print(batch_dot)
```

*What each operation does and why:*
- **Addition** (`batch_sum`) combined two 5-row batches into one running total per column — the same mechanic used to merge or aggregate separate data batches (Lipschutz & Lipson, 2018).
- **Subtraction** (`batch_diff`) captured how each person in `batch_a` differs from the matching person in `batch_b`, feature by feature — useful for directly comparing two groups or records (Lipschutz & Lipson, 2018).
- **Multiplication** confirmed multiplying by the identity matrix leaves data unchanged, while multiplying by our scaling matrix doubled only the `age` column — showing how a matrix can selectively rescale individual features (Lipschutz & Lipson, 2018; GeeksforGeeks, 2025).
- **Dot product** collapsed the entire `age` and `hours-per-week` columns into a single number (51,176,886), summarizing how those two features move together across all 32,561 people (Lipschutz & Lipson, 2018).

*Why this matters in AI:*
- **Data transformation** — combining and comparing batches (addition/subtraction) is the same basic operation used when merging mini-batches or measuring how a dataset shifts over time.
- **Feature scaling** — multiplying by a diagonal scaling matrix is literally how features get rescaled (e.g., normalization) before training, so no single large-range feature (like `fnlwgt`) dominates a model over a small-range one (like `age`).
- **Projections** — the dot product is the core operation behind similarity measures (like cosine similarity) and is exactly what a single neuron in a neural network computes: a weighted sum of its inputs.

## Part 2: Matrix Transpose and Inverses

### 1. Matrix Transpose

Transpose just flips a matrix — rows become columns. It's used constantly inside AI training (like gradient descent).

```python
print(f'Matrix: \n{X}')
print('\n')

trans_X = X.T
print(f'Matrix Transpose: \n{trans_X}')
```

### 2. Matrix Inverse

`X` itself isn't square, so it can't be inverted directly. We'll build `XᵀX` — multiplying `X` by its own transpose always gives a square matrix. This exact matrix (`XᵀX`) is the key ingredient in the "normal equation," the formula that solves linear regression directly.

Inverse is the "undo" of a matrix — like how dividing undoes multiplying for regular numbers. We'll invert `XᵀX` to see why that's needed to solve for regression weights.

```python
inverse_X = np.linalg.inv(trans_X @ X)
print(f'Matrix Inverse: \n{inverse_X}')
```

**Deliverables:**
- Code showing the matrix transpose and inverse computations.
- Discussion of why these operations are important in AI, particularly in optimization and equation solving.

**Answers:**

*What each operation does and why:*
- **Transpose** (`trans_X`) flips `X`'s rows into columns with `X.T`. We needed this because `X` is 32561×6 — not square — so it can't be inverted directly. Multiplying `X` by its transpose (`XᵀX`) gives us a square 6×6 matrix, which *can* be inverted (Lipschutz & Lipson, 2018).
- **Inverse** (`inverse_X`) is the "undo" of multiplication, the same way dividing undoes multiplying for regular numbers (`A × A⁻¹` gives the identity matrix, just like `5 × 2 ÷ 2` gets you back to 5) (Lipschutz & Lipson, 2018). We computed `np.linalg.inv(trans_X @ X)` to invert that square `XᵀX` matrix.

*Why this matters in AI:*
- **Equation solving** — `(XᵀX)⁻¹` is the key piece of the "normal equation," `weights = (XᵀX)⁻¹ Xᵀy`, which solves linear regression directly in one step (Rodriguez & Zieffler, 2025).
- **Optimization** — without being able to invert `XᵀX`, there'd be no direct formula for the best-fit weights. We'd have to fall back on slower, iterative methods like gradient descent instead of solving for the answer immediately.

## Part 3: Matrix Decomposition and Rank

### 1. Matrix Rank

Rank is a number that tells you how much real, independent information is in a matrix. If a matrix has 6 columns but a rank of only 4, it means 2 of those columns aren't adding anything new — they can be built from the others. In AI, low rank is a warning sign: some features may be redundant.

```python
rank = np.linalg.matrix_rank(X)
print(f'Matrix Rank: \n{rank}')
```

**Deliverables:**
- Code that calculates the matrix rank.
- Discussion of its implications for data analysis in AI.

**Answers:**

Rank tells you how many columns actually carry unique information (Lipschutz & Lipson, 2018). Since `X`'s rank is 6 (same as its column count), all 6 features are independent — no redundancy. If the rank had been lower, say 4, it would mean 2 columns could be built from the others, and those columns could be dropped without losing information — reducing redundancy, speeding up training, and preventing the model from wasting effort on repeated information (Rodriguez & Zieffler, 2025).

## Part 4: Vector Spaces and Linear Independence

### 1. Linear Independence Test

A set of columns is linearly independent if none of them can be written as a combination of the others — each one brings genuinely new information. We'll check this using rank again: if the rank equals the number of columns we're testing, they're independent. This matters in AI because feeding a model redundant features wastes effort and can confuse training.

```python
is_independent = X.shape[1] == rank
print(f'Number of columns: {X.shape[1]}')
print(f'Rank: {rank}')
print(f'Linearly independent: {is_independent}')
```

**Deliverables:**
- Code showing how you test for linear independence.
- Discussion of the importance of linear independence in feature selection for AI models.

**Answers:**

Linear independence matters for feature selection because redundant columns don't add new information — they just waste computation and can confuse a model during training without improving its accuracy (Lipschutz & Lipson, 2018; Rodriguez & Zieffler, 2025). In our case, since `X`'s columns are all independent (rank = number of columns = 6), we know all 6 features are worth keeping — none of them need to be removed.
